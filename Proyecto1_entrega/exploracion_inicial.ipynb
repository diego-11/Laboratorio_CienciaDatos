```python
# Separar el texto de TIPO DE HURTO en sus partes
df["ARTICULO_PENAL"] = df["TIPO DE HURTO"].str.extract(r"(ARTÍCULO \d+)")
df["TIPO_VEHICULO"] = df["TIPO DE HURTO"].str.extract(r"HURTO (\w+)$")

print(df["ARTICULO_PENAL"].unique())
print(df["TIPO_VEHICULO"].unique())
```
